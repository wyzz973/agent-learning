### 随身夹减小的是输入，不应抹掉完整依据

完整档案与本次模型上下文可以分开：档案保存原文，资料夹选择本次问题需要的片段或摘要，并附可取回的位置。裁剪只是删除字符，摘要则重新表达含义，两者都可能丢失关键日期、限制或冲突。Python切片不理解哪句重要，len计算字符数也不是精确token数，因此预算要明确统计范围，不能说一个短字符串就保证整个模型请求没有超限。

本人打包函数先保存原文及编号映射，再组织本次问题与摘要；取回函数只访问已登记且位于允许目录的路径。摘要中缺了信息时，阿灯需要真实调用工具拿到原文，不能把“路径可用”当成“已经读取”。相同内容的校验值有助核对取回版本，却不能证明它永远正确。聊天历史还有额外协议约束：工具申请与回执不可随意拆散，未完成动作也不能被剪没。本课资料pack并不自动成为支持任意历史的压缩器；迁移到长活动手册时仍要同时验预算、原文完整性和临时规则的支持关系。


## 1. 四种动作不是一回事
选择是决定本次带哪份资料；裁剪是删除部分已有内容；摘要是重新表达，可能丢失细节；外置是把完整内容保存在档案里，让模型之后按标识申请取回。
len(text) 测量 Python 字符串长度，不是 token 数。本关预算明确统计送入用户消息的 model_text 字符数，不包含系统提示与协议封套，不把它说成整个请求的精确 token 预算。
切片 text[:N] 只保留前 N 个字符，不理解哪一句是限制。先看一个不调用模型的小例子。

````python
import asyncio
import json
import os
import sys
import time
from pathlib import Path
from uuid import uuid4

from dotenv import load_dotenv
from IPython.display import Markdown, display
from langchain.chat_models import init_chat_model
from langchain.messages import AIMessage, HumanMessage, SystemMessage, ToolMessage

ROOT = Path.cwd().resolve()
for candidate in [ROOT] + list(ROOT.parents):
    if (candidate / "pyproject.toml").is_file() and (candidate / ".env").is_file():
        ROOT = candidate
        break
else:
    raise FileNotFoundError("请在图书馆项目目录中打开 Notebook")
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
load_dotenv(ROOT / ".env", override=False)
original_model = init_chat_model(os.environ["DEFAULT_MODEL"], timeout=20, max_retries=0)
MODEL = original_model.model_copy()
MODEL.extra_body = dict(original_model.extra_body or {})
MODEL.extra_body["thinking"] = {"type": "disabled"}
TASK_ID = "M04-T04"
PROMPT = (ROOT / "world/prompts" / (TASK_ID + ".md")).read_text(encoding="utf-8")
OUT = ROOT / "outputs/fog-island" / TASK_ID
OUT.mkdir(parents=True, exist_ok=True)
NOTICE_A = (ROOT / "world/notices/opening.md").read_text(encoding="utf-8")
NOTICE_C = (ROOT / "world/notices/return-box.md").read_text(encoding="utf-8")
display(Markdown("### 阿灯本关实际收到的岗位提示\n\n" + PROMPT))
````

## 2. 完整原文先归档，摘要与它分开
下面把 NOTICE-C 保存到教师档案，sha256 用于核对取回的是同一份内容，不是模型改写过的版本。文件内容来自 world，不把摘要覆盖回原文。
接着让阿灯生成一张简短便签。每段最多调用模型一次；这个便签仍需对照原文检查，不能因名字叫 summary 就视为正确。


````python
import hashlib

teacher_archive = OUT / "teacher-archive"
teacher_archive.mkdir(parents=True, exist_ok=True)
original_path = teacher_archive / "NOTICE-C.txt"
original_path.write_text(NOTICE_C, encoding="utf-8")
original_hash = hashlib.sha256(NOTICE_C.encode("utf-8")).hexdigest()
async with asyncio.timeout(60):
    summary_reply = await MODEL.ainvoke(
        [
            SystemMessage(PROMPT),
            HumanMessage("林禾请你为随身夹抄一张短便签，保留地点、使用条件和编号。\n" + NOTICE_C),
        ]
    )
print("原文字符数：", len(NOTICE_C), "便签字符数：", len(summary_reply.text))
display(Markdown(summary_reply.text))
assert original_path.read_text(encoding="utf-8") == NOTICE_C
````

## 3. 故意漏掉一条限制，看原文怎样回来
下面的 damaged_summary 是教师主动删减的便签，不伪称模型刚才一定漏写。它只保留馆内关闭，却没有保留还书箱信息。
先不给取回工具，看看阿灯会不会自然说明资料不足；再给出真实取回工具，观察文件是否真的被读取。即使基线回答碰巧正确，也不能证明它从缺失摘要里取得了依据。


````python
damaged_summary = "[NOTICE-C] 周日馆内关闭。"
reader_question = "周日还能还书吗？还书箱在哪？"
async with asyncio.timeout(60):
    without_fetch = await MODEL.ainvoke(
        [
            SystemMessage(PROMPT),
            HumanMessage(reader_question + "\n随身夹便签：" + damaged_summary),
        ]
    )
display(Markdown(without_fetch.text))
````

````python
from langchain.agents import create_agent
from langchain.agents.middleware import ModelCallLimitMiddleware
from langchain.tools import tool

fetch_log = []


@tool
async def read_original_notice(source_id: str) -> str:
    """到档案室取回随身夹便签对应的完整公告。
    Args:
        source_id: 当前可取 NOTICE-C；不能传任意文件路径。
    Returns:
        实际读取的原文。
    Raises:
        ValueError: 公告编号不在已登记档案中。
    """
    if source_id != "NOTICE-C":
        raise ValueError("这格档案只登记了NOTICE-C")
    text = original_path.read_text(encoding="utf-8")
    fetch_log.append(
        {"source_id": source_id, "sha256": hashlib.sha256(text.encode("utf-8")).hexdigest()}
    )
    return text


teacher_agent = create_agent(
    MODEL,
    tools=[read_original_notice],
    system_prompt=PROMPT,
    middleware=[ModelCallLimitMiddleware(run_limit=3, exit_behavior="error")],
)
````

````python
async with asyncio.timeout(60):
    with_fetch = await teacher_agent.ainvoke(
        {
            "messages": [
                HumanMessage(
                    reader_question
                    + "\n随身夹："
                    + damaged_summary
                    + "\n档案索引：NOTICE-C可以取回原文。"
                )
            ]
        },
        config={"recursion_limit": 10},
    )
assert fetch_log and fetch_log[0]["sha256"] == original_hash
display(Markdown(with_fetch["messages"][-1].text))
(OUT / "teacher-fetch-back.json").write_text(
    json.dumps(
        {
            "fetches": fetch_log,
            "without": without_fetch.text,
            "with": with_fetch["messages"][-1].text,
        },
        ensure_ascii=False,
        indent=2,
    ),
    encoding="utf-8",
)
````

**换输入迁移**：加入更长的活动手册，用同样策略选择和取回，不能靠无限增加上下文窗口解决。

先比较本次参数与实际模型输入，再检查执行回执和最后输出。定位第一处偏差，只改这一层；类型注解、框架调用成功和故事描述都不能替真实数据证明完成。


````python
from typing import Any

async def pack_context(question: str, documents: list[dict[str, str]],
                       archive_dir: Path, max_chars: int, model: Any = None,
                       role: str | None = None) -> dict:
    """完整归档原文，用一次摘要调用生成有预算的本轮输入。
    Args:
        question: 当前读者问题，不得丢失。
        documents: 含 id/title/text 的完整资料。
        archive_dir: 本关允许保存原文的位置。
        max_chars: model_text正字符数上限。model/role有值时必须使用它们。
        空documents零摘要；非空documents恰好通过传入模型调用一次。
    Returns:
        question、model_text、archives、chars。
    Raises:
        ValueError: 输入或预算不合法，摘要超出预算。
        NotImplementedError: 本人未完成。
    """
    raise NotImplementedError("请本人实现资料归档与上下文打包")

def fetch_archived(source_id: str, archives: dict[str, str], archive_dir: Path) -> dict:
    """只取回登记且位于允许目录中的完整原文。
    Args:
        source_id: 登记的来源编号。
        archives: 本人生成的编号到路径映射。
        archive_dir: 允许读取的档案根目录。
    Returns:
        source_id、text、sha256。
    Raises:
        ValueError: 未知编号或路径越界。
        NotImplementedError: 本人未完成。
    """
    raise NotImplementedError("请本人实现原文取回")
````

分层提示：Path.resolve 与 is_relative_to 用来检查目标仍在允许目录里；write_text 保存完整原文，不能保存摘要替代它。用 json.dumps 可以把资料元数据组织成可读文本，但长度检查应该针对最终 model_text，而不只算摘要一行。
先让一份资料归档与取回完全一致，再加入模型摘要；最后故意设极小预算，确认函数明确拒绝。不要用 Python 字符切片把无效摘要强行改成“成功”。
**可以休息。**回来先解释：原文在哪里，摘要在哪里，模型目前实际带了哪一份。


````python
from project.m02_tools import read_document
documents = []
for document_id in ["NOTICE-B", "VOL-01"]:
    receipt = await read_document.ainvoke({"document_id": document_id})
    if not receipt["ok"]:
        raise RuntimeError(receipt["error"])
    documents.append({"id": receipt["document_id"], "title": receipt["title"], "text": receipt["text"]})
archive_dir = OUT / "reader-archive"
question = "周六下午能来报名志愿者吗？最晚什么时候办？"
async with asyncio.timeout(60):
    context_pack = await pack_context(question, documents, archive_dir, 1200)
assert context_pack["chars"] == len(context_pack["model_text"]) <= 1200
assert context_pack["question"] == question
for document in documents:
    fetched = fetch_archived(document["id"], context_pack["archives"], archive_dir)
    assert fetched["text"] == document["text"]
(OUT / "context-pack.json").write_text(json.dumps(context_pack, ensure_ascii=False, indent=2), encoding="utf-8")
print("实际用户输入字符数：", context_pack["chars"])
print(context_pack["model_text"])
````

## 6. 同一个阿灯实际使用本人的资料夹与取回函数
包装工具只把模型请求交给你的 fetch_archived，不内置公告答案。通过一次真实取回记录，证明 archives 不只是写在纸上的路径。
本次 agent 最多调用模型 3 次，整段 60 秒；原文全文只由实际工具读取时返回，不把它偷偷预先塞进 system prompt。

工具标记 @tool 将 Python 函数提供给模型；create_agent 负责调用与回传循环，ModelCallLimitMiddleware 在这里把模型调用次数限制为三次。身份说明不会自动获得文件权限，读取仍必须经过你写的路径检查。


````python
student_fetches = []
@tool
async def fetch_from_folder(source_id: str) -> str:
    """按随身夹登记编号到档案室取原文。
    Args:
        source_id: 当前资料夹列出的编号。
    Returns:
        实际原文和来源编号。
    Raises:
        ValueError: 未登记或路径不允许。
    """
    fetched = fetch_archived(source_id, context_pack["archives"], archive_dir)
    student_fetches.append({"source_id": source_id, "sha256": fetched["sha256"]})
    return fetched["text"]
student_agent = create_agent(
    MODEL, tools=[fetch_from_folder], system_prompt=PROMPT,
    middleware=[ModelCallLimitMiddleware(run_limit=3, exit_behavior="error")],
)
async with asyncio.timeout(60):
    student_result = await student_agent.ainvoke(
        {"messages": [HumanMessage(context_pack["model_text"])]},
        config={"recursion_limit": 10},
    )
assert student_fetches, "需要实际取回记录，不能只声称原文可以读取"
display(Markdown(student_result["messages"][-1].text))
(OUT / "fetch-back-receipt.json").write_text(
    json.dumps({"fetches": student_fetches, "answer": student_result["messages"][-1].text},
               ensure_ascii=False, indent=2), encoding="utf-8",
)
````

### 验证预算失败与未登记来源

给出装不下问题的字符预算时，函数应在模型调用前拒绝，而非截断原问题。把未登记编号交给取回函数也应该明确拒绝。随后实际使用迁移资料夹发出一次模型请求；原始资料仍可从文件核对。


````python
try:
    await pack_context(question, documents, archive_dir, 1)
except ValueError:
    print("不可能的预算被拒绝，检查发生在调用之前的位置。")
else:
    raise AssertionError("不能截断问题来假装满足预算")
try:
    fetch_archived("UNREGISTERED", context_pack["archives"], archive_dir)
except ValueError:
    print("未登记来源被拒绝")
else:
    raise AssertionError("不能读未登记来源")
async with asyncio.timeout(60):
    transfer_reply = await MODEL.ainvoke([
        SystemMessage(PROMPT), HumanMessage(transfer_pack["model_text"])
    ])
display(Markdown(transfer_reply.text))
(OUT / "transfer-answer.md").write_text(transfer_reply.text, encoding="utf-8")
````
