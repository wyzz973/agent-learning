### 马上会遇到的专业词

**Reflexion**：把失败反馈整理成带适用条件的可复用经验。经验是否改善下一次行为必须做真实对照，不宣称训练了模型。

**Trace**：按顺序保存公开输入身份、动作、观察、状态和耗时，用来定位第一处可证偏差；不等于隐藏思维链。

**Ablation**：消融：保持其他条件尽量相同，移除一个机制观察变化。一次结果不足以建立普遍因果结论。

### 教训要带失败证据和适用条件

一次错误可以发生在来源选择、工具执行或理解原文时，先确定哪一层出现偏差，才知道该留下什么经验。程序漏送临时公告是可检查的输入缺口，模型是否因此答错则要读实际输出；不能为了展示反思而编造一次模型失误。经验记录保留失败检查、原始和更正依据、工作建议、版本与适用条件，不只是一个正确答案。

选择器依据当前咨询的条件读取仍启用的记录，再把它放入本次上下文；记录存在不等于已被使用。Python字典与过滤控制范围，active和version支持撤回与修订，新的记录不应修改原始失败轨迹。Reflexion式文本经验影响后续输入，不更新模型权重，也不保证一定改善。迁移到活动截止问题时，应重新查活动原文，不把旧开门时间带过去。有无经验要保持同题、同源与相同工具，保留实际差异；若两边都正确或经验反而误导，也应据实记录，而不往便条里塞入本题答案制造提升。


## 1. 经验与偏好、事实、训练有什么区别
读者偏好说明希望如何服务；公告说明某次安排；经验说明某种失败发生后，下次需要额外检查什么。它们都不是永久真理。
Reflexion 研究的一种思路是让外部任务反馈形成文本经验，在之后的尝试中使用，而不更新模型权重。本关只做应用级的小实验，不声称复现论文成绩，也不把一次改进说成阿灯普遍学会了所有规则。
先复现一个确定的应用错误：程序只把正式公告 NOTICE-A 送给模型，却漏掉已经在馆内生效的临时公告 NOTICE-B。错的是来源选择；模型具体会怎么说，以真实输出为准。

````python
import asyncio
import json
import os
import sys
import time
from pathlib import Path
from uuid import uuid4

from dotenv import load_dotenv
from IPython.display import Markdown, display
from langchain.chat_models import init_chat_model
from langchain.messages import AIMessage, HumanMessage, SystemMessage, ToolMessage

ROOT = Path.cwd().resolve()
for candidate in [ROOT] + list(ROOT.parents):
    if (candidate / "pyproject.toml").is_file() and (candidate / ".env").is_file():
        ROOT = candidate
        break
else:
    raise FileNotFoundError("请在图书馆项目目录中打开 Notebook")
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
load_dotenv(ROOT / ".env", override=False)
original_model = init_chat_model(os.environ["DEFAULT_MODEL"], timeout=20, max_retries=0)
MODEL = original_model.model_copy()
MODEL.extra_body = dict(original_model.extra_body or {})
MODEL.extra_body["thinking"] = {"type": "disabled"}
TASK_ID = "M04-T05"
PROMPT = (ROOT / "world/prompts" / (TASK_ID + ".md")).read_text(encoding="utf-8")
OUT = ROOT / "outputs/fog-island" / TASK_ID
OUT.mkdir(parents=True, exist_ok=True)
NOTICE_A = (ROOT / "world/notices/opening.md").read_text(encoding="utf-8")
NOTICE_C = (ROOT / "world/notices/return-box.md").read_text(encoding="utf-8")
display(Markdown("### 阿灯本关实际收到的岗位提示\n\n" + PROMPT))
````

## 2. 先有失败证据，再谈教训
这里的失败判据由馆务事实决定：处理这次临时调整应取得 NOTICE-B，但实际只给了 NOTICE-A。我们不是故意把模型输出写成错话，也不要求它一定幻觉。
下面只发一次真实模型请求，保存公开输入、答复和实际来源 ID。set 表示一组不重复的编号；difference 找出要求中的来源有哪些没被送入。它验证来源覆盖，不是自动理解答案含义。


````python
temporary_notice = (ROOT / "world/notices/temporary.md").read_text(encoding="utf-8")
question = "这周六几点可以进馆，几点闭馆？"
observed_ids = ["NOTICE-A"]
async with asyncio.timeout(60):
    old_reply = await MODEL.ainvoke(
        [
            SystemMessage(PROMPT),
            HumanMessage(question + "\n接待台送来的公告：\n" + NOTICE_A),
        ]
    )
required_ids = {"NOTICE-B"}
missing_ids = sorted(required_ids.difference(observed_ids))
failure_trace = {
    "trace_id": "notice-" + uuid4().hex,
    "question": question,
    "sources_sent": observed_ids,
    "public_answer": old_reply.text,
    "failed_check": "required_source_coverage",
    "missing_source_ids": missing_ids,
    "error_type": "temporary-update-missed",
    "scope": "temporary-schedule",
    "original_source": NOTICE_A,
    "correcting_source": temporary_notice,
}
assert missing_ids == ["NOTICE-B"]
display(Markdown(old_reply.text))
print("实际失败检查：", failure_trace["failed_check"], "未取得：", missing_ids)
(OUT / "failure-trace.json").write_text(
    json.dumps(failure_trace, ensure_ascii=False, indent=2), encoding="utf-8"
)
````

### 林禾的反馈具体指向哪里
如果答复采用旧时间，输出与临时安排不一致；如果答复自然表示不确定，来源选择仍然缺了一份必要材料。不要把两个问题混在一起，也不要把模型实际回答改写成预期错误。
下面把失败记录与更正公告交给阿灯，让它提出一个改进建议。模型建议不是自动生效的规则，你要对照实际缺口决定保留什么。它不能从一次开馆时间变化推断以后所有临时公告永远优先于任何文件。


````python
feedback_input = (
    "林禾核对了这次接待：临时公告没有送到你手里。请给她一句可执行的改进建议，"
    "再说清什么情况下应该用它；不要给今后的每份资料都套上同一个结论。\n"
    "接待记录：" + json.dumps(failure_trace, ensure_ascii=False)
)
async with asyncio.timeout(60):
    reflection_reply = await MODEL.ainvoke([SystemMessage(PROMPT), HumanMessage(feedback_input)])
display(Markdown(reflection_reply.text))
(OUT / "teacher-reflection-suggestion.md").write_text(reflection_reply.text, encoding="utf-8")
````

## 3. 修复一次输入，不等于形成可复用经验
下面直接补齐 A 与 B，再问相同问题，模型最多调用一次。你可以观察来源补齐怎样影响答复，但这一段并没有实现“下次自动找到合适经验”。
真正的经验机制需要你设计记录、选择与撤回。工作方法应指向去查适用范围和变更内容，而不是把“14:00”作为所有问题的固定答案。


````python
async with asyncio.timeout(60):
    corrected_reply = await MODEL.ainvoke(
        [
            SystemMessage(PROMPT),
            HumanMessage(
                question
                + "\n林禾补齐了这两张公告，请核对这次安排：\n"
                + NOTICE_A
                + "\n"
                + temporary_notice
            ),
        ]
    )
display(Markdown(corrected_reply.text))
(OUT / "teacher-source-correction.json").write_text(
    json.dumps(
        {
            "before": old_reply.text,
            "after": corrected_reply.text,
            "before_sources": observed_ids,
            "after_sources": ["NOTICE-A", "NOTICE-B"],
        },
        ensure_ascii=False,
        indent=2,
    ),
    encoding="utf-8",
)
````

## 4. 一张可核对的便条要带哪些字段
record_id 是稳定标识；version 是修订次数；active 控制是否继续使用。failure_trace_id、failed_check、evidence 保存教训由来；lesson 是你写的工作建议；conditions 说明适用范围。
条件匹配依据当前咨询的 error_type/scope 元数据，元数据由程序给出，不让模型自己选择另一个用户的经验。这个演示采用明确匹配规则，不把关键词筛选说成理解任意情境的智能检索。
dict(record) 只复制一层；嵌套对象仍可能共享。撤回本关记录只需改变顶层 active/version，所以可以用浅拷贝；不能据此声称所有深层内容都已复制。


````python
sample_note = {"record_id": "sample", "active": True, "version": 1}
changed_note = dict(sample_note)
changed_note["active"] = False
changed_note["version"] += 1
print("原记录：", sample_note)
print("新版本：", changed_note)
assert sample_note["active"] is True
````

## 6. 新活动委托：同一种错误，不同的答案
活动手册 EVENT-01 有原定截止时间，EVENT-02 只改这次活动的截止安排。它们来自 world/events，不是旧开馆公告的重命名。
下面工具允许阿灯按编号取两份活动原文。无经验与有经验使用同一个问题、目录、模型与工具；差别只有你的选择器是否把便条送入上下文。即使两种都答对，也要诚实记录，不人为降低基线。
便条中保存的旧馆务数字只是失败背景，不能覆盖新活动原文。检查最终引用与真实工具回执，不要求模型公开内部推理。


````python
from langchain.agents import create_agent
from langchain.agents.middleware import ModelCallLimitMiddleware
from langchain.tools import tool

EVENTS = {
    "EVENT-01": (ROOT / "world/events/reading-club.md").read_text(encoding="utf-8"),
    "EVENT-02": (ROOT / "world/events/reading-club-update.md").read_text(encoding="utf-8"),
}


@tool
async def read_event_notice(source_id: str) -> str:
    """为阿灯取出读书活动安排或临时调整。
    Args:
        source_id: EVENT-01为原安排，EVENT-02为这次活动的临时调整。
    Returns:
        实际读取的完整活动公告。
    Raises:
        ValueError: 编号不在活动档案中。
    """
    if source_id not in EVENTS:
        raise ValueError("这格只存了EVENT-01和EVENT-02")
    return EVENTS[source_id]
````

### 对照设施只消费你的便条，不替你选择
_run_event_case 接收经验列表，将它与新读者问题一起交给阿灯。它不实现 choose_experiences，也不会在没有便条时偷用教师规则。
每个条件最多调用模型 4 次，单次 20 秒，整段 60 秒。两个条件分别有边界，避免某一条件的耗时挤占另一条件的预算。事件轨迹只保存正文、工具参数、配对编号和次数。


````python
async def _run_event_case(question: str, notes: list[dict]) -> dict:
    agent = create_agent(
        MODEL,
        tools=[read_event_notice],
        system_prompt=PROMPT,
        middleware=[ModelCallLimitMiddleware(run_limit=4, exit_behavior="error")],
    )
    request = (
        question
        + "\n活动资料：EVENT-01原安排；EVENT-02临时调整。"
        + "\n可参考的纠错便条："
        + json.dumps(notes, ensure_ascii=False)
    )
    async with asyncio.timeout(60):
        result = await agent.ainvoke(
            {"messages": [HumanMessage(request)]}, config={"recursion_limit": 14}
        )
    public_trace = []
    calls = 0
    for message in result["messages"]:
        row = {"role": message.type, "text": message.text}
        if isinstance(message, AIMessage):
            row["tool_calls"] = message.tool_calls
            calls += 1
        if isinstance(message, ToolMessage):
            row["tool_call_id"] = message.tool_call_id
        public_trace.append(row)
    return {
        "answer": result["messages"][-1].text,
        "trace": public_trace,
        "model_calls": calls,
        "experience_ids": [note["record_id"] for note in notes],
    }
````

````python
event_question = "这次读书活动最晚什么时候报名？以后每场活动也都按这个时间吗？"
without_experience = await _run_event_case(event_question, [])
applicable = choose_experiences(reloaded_notes, matching_context)
with_experience = await _run_event_case(event_question, applicable)
assert with_experience["experience_ids"] == [experience["record_id"]]
assert without_experience["experience_ids"] == []
print("没有便条：")
display(Markdown(without_experience["answer"]))
print("使用本人便条：")
display(Markdown(with_experience["answer"]))
comparison = {"question": event_question, "without": without_experience,
              "with": with_experience, "scope": "一次真实对照，不代表普遍提升"}
(OUT / "with-without-experience.json").write_text(json.dumps(comparison, ensure_ascii=False, indent=2), encoding="utf-8")
````
