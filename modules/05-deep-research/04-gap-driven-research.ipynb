### 马上会遇到的专业词

**研究**：围绕明确问题与完成条件发现来源、取得原文、评估缺口并有限继续。搜索摘要不是已经读取的原文。

**工具循环**：模型生成动作→程序执行→观察进入下一次请求，反复到完成或停止。只有一次模型调用且没有环境反馈时，未必需要agent循环。

**预算**：分别约束调用尝试、图步数、工具动作、时间与输入大小；失败尝试也可能消耗资源。

**RAG**：Retrieval-augmented generation，先检索当前证据，再让模型基于它生成。检索召回、来源可靠与生成有据需要分别验证。

### 缺口检查必须改变下一步，而不只写一句批评

有若干来源不等于已经回答全部研究要求。研究状态需要把原子题、已有证据、发现的冲突和未解决项分别保存，检查器才能指出哪部分仍没有依据。结构化检查结果进入条件路由，决定继续检索、请林禾澄清，或在证据足够时形成报告。模型生成一段“需要更深入研究”的评论，如果没有改变后续动作，就没有形成缺口驱动循环。

Python集合差集可以表达覆盖缺项，状态更新保留新增证据，有限轮数给持续搜索设边界。查到更多相似段落不必然补上原缺口，查询应针对具体未满足问题；不同版本的官方说法也不能随意合成一句无条件结论。预算耗尽时要保留仍未知与冲突，不把停止当作答案齐全。本人迁移时保持原研究图，换成有版本差异的资料，检查路由选择是否随实际证据变化。报告要能指向当前来源窗口，已有文件、长篇文字和模型自报完成都不能代替这条证据链。

## 4. 本人核心：完整连接有界研究图

你写两个函数，不能直接返回教师状态：

- `next_research_step(state)`：缺口为空且还有写作调用时返回`write`；缺口仍在、研究轮数和模型预算允许继续时返回`research`；否则返回`stop`。为下一次检查和可能的写作留出调用额度；有来源时每个节点需要摘要与研究两次，无来源时本次判断只需一次，不能只数网页。
- `build_gap_graph(state_schema, nodes, checkpointer=None)`：注册assess/research/write/stop；开始进入assess；assess的条件边调用本人路由；research回到assess；write与stop结束。compile时传入checkpointer，后续服务才能复用本人图恢复。

`nodes`保存节点函数对象，注册时不加括号调用。`checkpointer=None`是可选参数，表示本次不保存跨进程状态，并非所有研究都天然可恢复。

本人初始取证会使用上一关的真实索引与本人排序，随后才按缺口决定是否查新网页。

````python
context_adapter = lab.owned_context(ROOT, model, SYSTEM_PROMPT, OWN / "context-archives")
owned_nodes = lab.make_gap_nodes(model, SYSTEM_PROMPT, context=context_adapter,
                                 search=lab.search_official_collection)
owned_graph = build_gap_graph(lab.GapState, owned_nodes)
initial_state = {"question": index["brief"]["question"], "sources": seed_sources, "brief": index["brief"],
    "missing": [], "query": "", "rounds": 0, "max_rounds": 2, "calls": 0,
    "max_calls": 8, "context_calls": 0, "reasoning_calls": 0, "answer": "", "status": "running", "events": []}
async with asyncio.timeout(180):
    result = await owned_graph.ainvoke(initial_state, {"recursion_limit": 30})
assert result["brief"] == index["brief"], "完整简报必须留在实际研究状态中"
assert result["calls"] <= 8 and result["rounds"] <= 2
assert result["calls"] == result["context_calls"] + result["reasoning_calls"]
assert result["status"] in {"reported", "needs_evidence"}
research_events = [event for event in result["events"] if event["stage"] == "research"]
print("实际追加研究次数：", len(research_events), "终止状态：", result["status"])
(OWN / "research-report.md").write_text(result["answer"], encoding="utf-8")
display(Markdown(result["answer"]))
research_run = {"brief": index["brief"], "result": result, "upstream": index_record["run_dir"]}
````
