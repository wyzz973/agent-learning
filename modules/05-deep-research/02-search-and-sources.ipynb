`re.findall`从Markdown链接中提取标题和URL；`list`保留候选，`.sort(key=...)`决定排名。教师设施里的排序规则可查看，但本关你主要负责由本人简报驱动完整取证，不花整关重写正则。

````python
async with asyncio.timeout(90):
    demo_hits = await lab.search_official("persistence", 2)
    if not demo_hits:
        raise RuntimeError("官方索引没有候选，请核查查询和站点状态")
    print("本次实际发现：", demo_hits)
    demo_source = await lab.fetch_public(demo_hits[0]["url"])
    assert demo_source["ok"], demo_source
    demo_answer = await lab.ask(
        model,
        SYSTEM_PROMPT,
        "林禾想知道检查点大致解决什么问题。请根据取回的这段官方原文简短说明并附URL。\n"
        + lab.snapshot_input([demo_source]),
    )
lab.save_json(DEMO / "discovered-sources.json", {"hits": demo_hits, "source": demo_source})
(DEMO / "source-note.md").write_text(demo_answer, encoding="utf-8")
display(Markdown(demo_answer))
````
