## 原理一：协作任务要有可合并的输出契约

委托每个研究单元时明确子题、来源范围、输出形状、预算和停止条件。否则它们会重复查询、留下缺口或把整体目标误当各自任务。独立任务可并发，存在依赖时固定顺序；分工开销要进入总成本。

结果包含真实sources、状态、失败、调用与未覆盖项。合并按URL与内容身份，保留冲突版本；最终writer只消费成功取得的证据，失败标成部分结果。不能以任务列表空了或所有协程返回就宣称研究完成。

## 原理二：评估单位包含环境和轨迹

一个case不只有prompt，还包括材料版本、初始状态、身份/权限、允许工具、预算和预期检查。run产生输出、动作、环境变化与轨迹，grader核对目标和边界。单测检查组件，端到端检查输入到产物，生产观测看真实分布，各自有证据范围。

开发集用于改实现，回归集保护已有能力，留出集在候选与方法锁定后验收。reference/expected留在评分侧，不能给被测模型。不同的合法工具路径可能达成同一目标，检查结果与关键约束，避免把一种固定策略写成唯一解。

## 原理三：代码检查、语义检查与人工校准

代码适合类型、调用数、权限、来源身份、测试SHA等明确规则；语义裁判辅助判断结论支持与完整性。裁判可能受顺序、篇幅、措辞影响，需盲化、反转顺序、与人工校准样本比较。

置信不足和裁判不一致保留needs_review。原文短句存在不自动证明支持关系，也不保存隐藏推理。一次好分或一条漂亮理由不能代替原文核对。比较两个方案时保持输入与环境，记录模型/prompt/依赖版本，并保留所有失败样本。

## 原理四：trace是一条跨层因果链

request_id关联业务请求，trace_id关联执行轨迹，span_id关联一次具体操作，parent/link表示同步或异步因果关系。记录开始/结束、状态、延迟、尝试、输入/来源指纹；日志不是任意打印所有消息。

先找首个与契约不符的位置：来源没有取得、取得没打包、打包没发送、观察错配、恢复用了旧身份，或最终生成扩大范围。LLM层、工具层、存储层和协议层分别标注。OpenTelemetry是可观测信号的标准，课程本地JSON轨迹用于学习结构，不把自定义日志冒称已经部署OTLP系统。

## 原理五：质量、成本和可靠性需共同解释

成功率、拒答恰当性、证据覆盖、重复运行差异、p50/p95延迟、实际调用和未知成本分别报告。小样本不能支撑精确稳定的p95或推广结论。故障由基础设施造成时，要与模型能力分开分析，环境资源和并发配置也写进实验身份。

关键边界失败优先于平均高分。越权、伪造发布、引用伪造、测试版本错配应阻止相关交付；可接受的unknown或取消不能按模型失败胡乱扣分。SLO由本章业务情景设定，实训阈值不是行业统一标准。

### 先看清本章原理

协作、评估与服务都依赖已经明确的研究契约。并行只适合独立或已满足依赖的子题，子agent隔离上下文能减少干扰，也可能漏掉汇合需要的证据；一个子任务失败不能自动被平均成整体成功。评估先固定输入、期望和检查方法，再对照基线与升级，程序性字段检查与语义判断分开；一次分数不是普遍质量保证。轨迹用标识和原文版本连接输入、工具、状态与结果，诊断应找到最早可证实的偏差，而不猜测看不到的内部推理。HTTP提交成功只说明请求被接收，任务完成需要持久状态与产物；重启后续做还要证明没有重做或串用其他会话。教学中要同时保留成功、失败和资料不足的真实样本，让本人先预测应检查哪项契约，再用统一标准判断。新增协作和服务入口不能悄悄换掉研究问题，否则无法解释系统升级的实际贡献。


### 分工先看依赖，汇合时带回依据

研究的几个子题可能互不依赖，也可能要先知道选型结果才能继续。并发适合前者，不会自动解决后者；把尚未获得的前提分配给多个研究者，只会让他们各自猜测。子agent的独立上下文能隔离大量过程，但返回摘要若漏掉来源或不确定性，主任务就失去检查条件。分工接口必须带回问题身份、证据与完成状态，不只是一段顺口结论。

Python异步协程在等待时允许其他工作推进，gather汇合结果，Semaphore限制同时进行的任务数；并发数与每个研究者的模型调用上限需要分开。一个worker失败时应保留该子题缺口，不能因其他结果成功就抹掉它。本人将同一Brief中的子题交给已经完成的研究器，主任务再依据真实返回汇总。固定并行流程和由模型动态委派是不同控制方式，都可以使用LangGraph，不要仅凭多个对象就声称自主协作。迁移增加依赖子题时，先前结果应真正成为新输入，再比较耗时与证据完整性。

````python
from langchain.agents import create_agent
from langchain.agents.middleware import ModelCallLimitMiddleware, ToolCallLimitMiddleware
from langchain.tools import tool

async with asyncio.timeout(60):
    hits = await lab.search_official("persistence", 1)
    assert hits
    team_source = await lab.fetch_public(hits[0]["url"])
assert team_source["ok"], team_source


@tool
async def read_recovery_page() -> str:
    """为阿灯研究模块读取已取回的官方恢复资料，保留URL和原文窗口。

    Args:
        无：只提供当前已取得的恢复页面。
    Returns:
        实际来源与原文窗口的JSON文本。
    Raises:
        无。
    """
    return lab.snapshot_input([team_source])
````

````python
async def narrow_worker(question: str) -> dict:
    """运行只使用一份恢复资料的窄研究模块。

    Args:
        question: 林禾的恢复子问题。
    Returns:
        真实答复、工具观察和模型调用数量。
    Raises:
        RuntimeError: 未观察到取证；预算或模型错误保留。
    """
    agent = create_agent(
        model,
        [read_recovery_page],
        system_prompt=SYSTEM_PROMPT
        + "\n你是阿灯的恢复研究模块，先读给定工具再回答当前子题，附URL。",
        middleware=[
            ModelCallLimitMiddleware(run_limit=2, exit_behavior="error"),
            ToolCallLimitMiddleware(run_limit=1, exit_behavior="error"),
        ],
    )
    started = time.monotonic()
    result = await agent.ainvoke({"messages": [HumanMessage(question)]}, {"recursion_limit": 40})
    observations = [m.text for m in result["messages"] if isinstance(m, ToolMessage)]
    if not observations:
        raise RuntimeError("研究模块没有取得工具观察")
    return {
        "question": question,
        "answer": result["messages"][-1].text,
        "observations": observations,
        "calls": sum(isinstance(m, AIMessage) for m in result["messages"]),
        "elapsed": time.monotonic() - started,
    }
````

````python
demo_questions = [
    "林禾想知道检查点保存什么，请附来源。",
    "林禾想知道thread_id为什么关系到恢复，请附来源。",
]
async with asyncio.timeout(120):
    demo_single = await narrow_worker("；".join(demo_questions))
    parallel_started = time.monotonic()
    demo_parallel = await asyncio.gather(
        narrow_worker(demo_questions[0]), narrow_worker(demo_questions[1])
    )
    parallel_elapsed = time.monotonic() - parallel_started
print("单研究者：", demo_single["calls"], "次模型，", round(demo_single["elapsed"], 2), "秒")
print(
    "两个模块：",
    sum(r["calls"] for r in demo_parallel),
    "次模型，",
    round(parallel_elapsed, 2),
    "秒",
)
lab.save_json(
    DEMO / "team-comparison.json",
    {"single": demo_single, "parallel": demo_parallel, "parallel_elapsed": parallel_elapsed},
)
display(Markdown(demo_single["answer"]))
````

## 4. 本人核心：有边界地派发并收回任务

`run_workers(questions, worker, max_workers=2)`接收两个由原brief产生的子问题。本人用Semaphore限制并发、用gather等待、保留原问题顺序。每项成功保留worker的真实状态；普通子任务失败返回含question/status="failed"/error_type的对象，其他任务仍可收回。未实现的NotImplementedError原样抛出，不能装成有意义的失败实验。

## 这份示范具体怎样工作

串行处理独立子问题会慢，但并行有协调成本：asyncio.gather 只保证按传入次序返回，不检查任务间依赖；盲目并行会浪费模型调用，有依赖时还会让后一步猜答案。限制并发（Semaphore）和调用预算（Middleware）让并行保留收益而不失控。若子任务必须顺序依赖或资源极少，就不该用并行委派。

## 把分工真正带回同一份研究作品

run_workers返回的每个研究者可能发现了新来源。只把分工前shared_sources保存到team-run，会让这些新证据在服务调用中丢失。本人实现merge_worker_sources(seed_sources, results)：

- 收集种子和每个成功/不足结果中的实际sources，失败模块没有伪造来源。
- 同URL、同原文SHA去重；SHA按实际text核对，不能只信字段。
- 同URL、不同正文保留两个版本，给不同版本分配不冲突的source_id，不能静默取最后一份。
- 返回sources和conflicts；conflicts说明URL、各版本source_id/sha256。这是版本冲突记录，不让程序假装判出了事实真伪。

后续报告和服务使用本人合并结果，原始字典不能被修改。

````python
def merge_worker_sources(seed_sources: list[dict], worker_results: list[dict]) -> dict:
    """把阿灯分工实际取得的来源合回总账，保留冲突版本。

    Args:
        seed_sources: 分工前来源；worker_results: 各模块真实结果或失败记录。
    Returns:
        sources和conflicts；同URL同SHA去重，不同SHA保留独立身份。
    Raises:
        ValueError: 来源正文/指纹不一致；NotImplementedError: 本人未完成。
    """
    raise NotImplementedError("请本人实现证据合并、去重与冲突保留")
````

````python
import hashlib

a = {"ok": True, "source_id": "same", "url": "https://docs.python.org/3/library/asyncio.html",
     "text": "原版本", "sha256": hashlib.sha256("原版本".encode()).hexdigest()}
b = {**a, "text": "修改版本", "sha256": hashlib.sha256("修改版本".encode()).hexdigest()}
combined = merge_worker_sources([a], [{"status": "reported", "sources": [dict(a), b]}])
assert len(combined["sources"]) == 2 and len(combined["conflicts"]) == 1
assert len({s["source_id"] for s in combined["sources"]}) == 2
assert {s["text"] for s in combined["sources"]} == {"原版本", "修改版本"}
assert a["source_id"] == "same" and b["source_id"] == "same", "不能修改上游记录"
print("这是合并规则的受控数据检查；正式运行还必须合并真实worker新来源。")
````
