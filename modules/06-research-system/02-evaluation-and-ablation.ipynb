## 3. 本人定义通过标准

`grade_run(result, case)`返回`{"passed":bool,"checks":{...}}`，至少检查：

- `status`：符合样例要求，资料不足不能伪报reported。
- `budget`：模型总调用在0～case.max_calls之间（默认8，包含摘要），追加轮数不越过样例上限。
- `source_trace`：reported必须有实际成功来源，技术题答复出现其中至少一个URL；本地馆务来源world:///则核对对应source_id通知编号；没有来源的报告不能过关。
- `completion`：reported有非空答复；needs_evidence保留非空missing与不足说明。

这些规则只验证执行和来源链，不能判定中文结论真的受到原文支持。另记录语义审读选择：支持/不支持/无法判断，逐项指向原文，不计入规则通过率。服务也会调用这个本人grader作为发布前执行检查，max_calls由本次完整pipeline预算给出。

全部检查满足才passed。不要调用模型打分；语义审读另留记录。未实现要抛错。函数不能把所有失败都算成功，也不能为比较好看删除样例。
