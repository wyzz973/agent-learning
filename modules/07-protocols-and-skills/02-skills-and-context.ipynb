````python
def public_trace(messages: list[AnyMessage]) -> list[dict[str, Any]]:
    """提取公开消息，完整AIMessage仍只在运行内存使用。

    Args:
        messages: 本次实际消息列表。
    Returns:
        角色、公开正文、工具调用及关联id，不含隐藏推理或请求元数据。
    Raises:
        无：只读取已知字段。
    """
    rows = []
    for message in messages:
        row: dict[str, Any] = {"role": message.type, "text": message.text}
        if isinstance(message, AIMessage):
            row["tool_calls"] = message.tool_calls
        if isinstance(message, ToolMessage):
            row["tool_call_id"] = message.tool_call_id
            row["tool_name"] = message.name
            row["status"] = message.status
        rows.append(row)
    return rows
````

````python
def save_record(name: str, payload: dict[str, Any]) -> Path:
    """保存本次实际记录，不推断学习者已掌握。

    Args:
        name: 当前目录中的JSON文件名，不允许路径片段。
        payload: 可序列化的公开数据，不包含密钥或隐藏推理。
    Returns:
        写入文件路径。
    Raises:
        ValueError: 文件名越界；OSError: 写入失败。
    """
    if Path(name).name != name or not name.endswith(".json"):
        raise ValueError("只允许当前作品目录中的JSON文件名")
    path = RUN_DIR / name
    path.write_text(json.dumps(payload, ensure_ascii=False, indent=2), encoding="utf-8")
    return path
````

## 1. 为什么先发现，再加载

Skill目录至少有SKILL.md，开头是YAML元数据，正文是方法，可附references、scripts和assets。启动时先看name与description，匹配任务后才读正文，需要时再取参考文件。这叫渐进披露，不等于把所有手册一次塞进prompt。

本页不执行Skill脚本，也不把allowed-tools字段当作已经获得权限。下面用yaml.safe_load解析元数据；它读取数据，不运行YAML中的代码。路径只来自当前手册架，不能让模型拼任意文件路径。


````python
def discover_methods(directory: Path) -> list[dict[str, Any]]:
    """发现手册元数据，不加载正文到模型上下文。

    Args:
        directory: 当前允许的Skill目录。
    Returns:
        名称、描述、真实文件路径；正文仍留在磁盘。
    Raises:
        ValueError: 目录、元数据或路径不合法；OSError: 读取失败。
    """
    found = []
    for path in sorted(directory.glob("*/SKILL.md")):
        if not path.resolve().is_relative_to(directory.resolve()):
            raise ValueError("手册路径越出允许目录")
        sections = path.read_text(encoding="utf-8").split("---", 2)
        if len(sections) != 3 or sections[0].strip():
            raise ValueError("SKILL.md需要YAML frontmatter")
        meta = yaml.safe_load(sections[1])
        if meta["name"] != path.parent.name or not meta["description"]:
            raise ValueError("手册名称需与目录一致，描述不能为空")
        found.append({"name": meta["name"], "description": meta["description"], "path": path})
    return found


methods = discover_methods(TEACHER_SKILLS)
metadata_only = [{"name": x["name"], "description": x["description"]} for x in methods]
print(metadata_only)
````

观察metadata_only：没有METHOD_BODY。Python已为解析读过文件，不等于模型已经收到正文；关键是实际发入消息的内容。后面的加载记录区分“文件被扫描”和“正文进入模型上下文”。

教师示范只有一册窄方法。让真实模型在这册和none之间选择；none是合法结果，不为了展示Skills而强行加载。


````python
class MethodChoice(BaseModel):
    """阿灯本次需要翻开的工作手册。"""

    name: Literal["return-box-review", "none"] = Field(description="有合适方法才选择，否则none")


selector = model.with_structured_output(MethodChoice, method="function_calling")
review_question = "帮林禾核对这句给读者的指引：周日馆内开放，请去西门还书箱还书。[NOTICE-C]"
async with asyncio.timeout(90):
    selected = await selector.ainvoke(
        [
            SystemMessage(SYSTEM_PROMPT + "\n先只根据方法目录选一本适用手册，没有适用的就选none。"),
            HumanMessage(
                review_question + "\n方法目录：" + json.dumps(metadata_only, ensure_ascii=False)
            ),
        ]
    )
print("真实选择：", selected.name)
assert selected.name == "return-box-review", "方法未匹配，不替模型篡改选择。"
loaded_method = methods[0]["path"].read_text(encoding="utf-8")
print("进入上下文的文件：", methods[0]["path"].name)
````

````python
@tool
def read_return_box() -> str:
    """领取雾岛图书馆东门还书箱的当前说明，核对地点和可使用时段。"""
    return (ROOT / "world/notices/return-box.md").read_text(encoding="utf-8")


app = create_agent(
    model,
    tools=[read_return_box],
    system_prompt=SYSTEM_PROMPT + "\n林禾交给你的工作手册：\n" + loaded_method,
    middleware=[ModelCallLimitMiddleware(run_limit=4, exit_behavior="error")],
)
async with asyncio.timeout(90):
    reviewed = await app.ainvoke({"messages": [HumanMessage(review_question)]})
display(Markdown(reviewed["messages"][-1].text))
save_record(
    "teacher-skill-use.json",
    {
        "question": review_question,
        "selected": selected.name,
        "loaded_files": [str(methods[0]["path"].relative_to(RUN_DIR))],
        "method_sha": hashlib.sha256(loaded_method.encode()).hexdigest(),
        "trace": public_trace(reviewed["messages"]),
    },
)
````

请核对两处错误：西门是否有原文支持？还书箱可用是否等于馆内开放？方法提供核对步骤，实际事实仍来自工具正文；模型读过手册也可能出错，需要检查输出。

下面只问一句你好。先预测是否还需要加载核查正文，再看实际选择。


````python
async with asyncio.timeout(45):
    greeting_choice = await selector.ainvoke(
        [
            SystemMessage(SYSTEM_PROMPT + "\n只选择适用手册；普通问候不需要引用核查，选none。"),
            HumanMessage(
                "你好，阿灯！\n方法目录：" + json.dumps(metadata_only, ensure_ascii=False)
            ),
        ]
    )
print("问候时的选择：", greeting_choice.name)
assert greeting_choice.name == "none", "若仍加载方法，先检查选择要求与描述。"
print("本次没有把手册正文发给模型。")
````

## 这份示范具体怎样工作

如果每次把方法全文塞进所有提示，上下文会被无关步骤占满，成本上升且容易走偏。Skill 把任务知识、步骤和参考资源分开保存，先只暴露方法目录的元数据，让模型按问题相关性挑一本，再只把被选中的 SKILL.md 正文装进上下文。这样只有真正需要时才有加载，普通问候或无关问题就选 none。若任务总处于同一领域、手册只有一份且模型必须每次都读，则直接全文放入也可行。

### 本人实现按需使用

answer_using_method必须先发现元数据，再决定是否加载正文；只读匹配的方法，不把整个目录全部拼进消息。选择可以使用结构化模型输出，但必须验证选择属于实际目录。普通问候允许不加载。

函数用本人的tools，保留当前情景SYSTEM_PROMPT，工具回答来自实际原文。最多一次选择调用加四次agent调用，整段90秒。返回selected（名称或None）、loaded_files、messages；不能返回教师reviewed。在结果中保留实际加载文件，便于证明选择影响了上下文。


````python
async def answer_using_method(question: str, directory: Path, tools: list[BaseTool]) -> dict[str, Any]:
    """本人按需选择本馆方法，再用实际工具回答。

    Args:
        question: 本轮访客请求；directory: 允许的Skill架；tools: 本人工具。
    Returns:
        selected、loaded_files、messages，来自本次真实运行。
    Raises:
        ValueError: 方法选择不合法；TimeoutError: 整段超过90秒；NotImplementedError: 未完成。
    """
    raise NotImplementedError("请本人实现发现、按需加载与真实使用。")

my_question = "请核对：下雨就不开馆，应该从西侧进门。[RAIN-01]"
my_result = await answer_using_method(my_question, MY_SKILLS, [read_document])
````
