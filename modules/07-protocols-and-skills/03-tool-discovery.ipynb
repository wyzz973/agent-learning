### 选择工具也是会出错的一次检索

工具很多时，把全部schema送进每次调用会增加输入负担，也让相近能力更容易混淆。能力目录先提供名称与简要说明，选择阶段判断本次可能需要的工具，加载阶段再交出实际schema与可执行对象。目录选择不等于动作已执行，更不等于选得少就一定更好；漏掉必要能力会让后面的agent无论怎样推理都取不到资料。

本人选择器消费当前问题和实际目录，输出目录内、不重复且符合数量限制的名称；执行层只绑定这个允许集合，保留没有命中或能力缺失的情况。Python字典、集合和过滤用于核验选择范围，模型负责判断相关性，两者都需要观察。对照应计算选择阶段自身的调用与耗时，再比较后续输入规模、成功取件和答复质量，不能只展示更小的工具列表。修改工具说明或新增能力后，要重新构造目录并用未见问题验证，旧选择缓存可能已经失效。MCP目录或提供方原生tool search只是可用连接方式，各自支持边界仍要核对。

````python
ROOT = Path.cwd()
for candidate in [ROOT] + list(ROOT.parents):
    if (candidate / "pyproject.toml").is_file() and (candidate / ".env").is_file():
        ROOT = candidate
        break
else:
    raise FileNotFoundError("请导师确认项目内核与.env。")
load_dotenv(ROOT / ".env", override=False)
MODEL_ID = os.environ["DEFAULT_MODEL"]
options: dict[str, Any] = {"temperature": 0, "timeout": 20, "max_retries": 0}
if MODEL_ID.startswith("deepseek:"):
    options["extra_body"] = {"thinking": {"type": "disabled"}}
model = init_chat_model(MODEL_ID, **options)
SYSTEM_PROMPT = (ROOT / "world/prompts" / (TASK_ID + ".md")).read_text(encoding="utf-8")
CATALOG = json.loads((ROOT / "world/cabinet/catalog.json").read_text(encoding="utf-8"))
RUN_DIR = ROOT / "outputs/fog-island" / TASK_ID / uuid.uuid4().hex[:12]
RUN_DIR.mkdir(parents=True, exist_ok=False)
print("本次作品目录：", RUN_DIR)
print(SYSTEM_PROMPT)
````

````python
def public_trace(messages: list[AnyMessage]) -> list[dict[str, Any]]:
    """提取公开消息，完整AIMessage仍只在运行内存使用。

    Args:
        messages: 本次实际消息列表。
    Returns:
        角色、公开正文、工具调用及关联id，不含隐藏推理或请求元数据。
    Raises:
        无：只读取已知字段。
    """
    rows = []
    for message in messages:
        row: dict[str, Any] = {"role": message.type, "text": message.text}
        if isinstance(message, AIMessage):
            row["tool_calls"] = message.tool_calls
        if isinstance(message, ToolMessage):
            row["tool_call_id"] = message.tool_call_id
            row["tool_name"] = message.name
            row["status"] = message.status
        rows.append(row)
    return rows
````

````python
def save_record(name: str, payload: dict[str, Any]) -> Path:
    """保存本次实际记录，不推断学习者已掌握。

    Args:
        name: 当前目录中的JSON文件名，不允许路径片段。
        payload: 可序列化的公开数据，不包含密钥或隐藏推理。
    Returns:
        写入文件路径。
    Raises:
        ValueError: 文件名越界；OSError: 写入失败。
    """
    if Path(name).name != name or not name.endswith(".json"):
        raise ValueError("只允许当前作品目录中的JSON文件名")
    path = RUN_DIR / name
    path.write_text(json.dumps(payload, ensure_ascii=False, indent=2), encoding="utf-8")
    return path
````

````python
from time import perf_counter

from pydantic import BaseModel, Field

LOOKUPS = [
    ("lookup_opening", "领取本周临时开馆时间，核对能否在某时到馆。", "NOTICE-B"),
    ("lookup_volunteers", "领取志愿者报名地点、时间和提交事项。", "VOL-01"),
    ("lookup_returns", "领取周日还书与还书箱位置的说明。", "NOTICE-C"),
    ("lookup_rain", "领取雨天入口和通行说明。", "RAIN-01"),
]


@tool
async def demo_document(document_id: str) -> dict[str, Any]:
    """按资料编号领取雾岛馆务原文，没找到时明确说明。"""
    for entry in CATALOG:
        if entry["id"] == document_id:
            return {
                "ok": True,
                "document_id": document_id,
                "text": (ROOT / entry["path"]).read_text(encoding="utf-8"),
            }
    return {"ok": False, "document_id": document_id, "error": "目录中没有这份资料。"}
````

## 1. 一个目录、四个有边界的入口

这些入口读取不同馆务资料，底层仍由同一个资料工具执行。下面用闭包把source_id记在各工具中，模型只看名称与用途。最后返回的是工具对象，不是已经读取的正文。

这一层是环境设施。本人部分要决定挑哪些工具、怎么验证选择、怎样让实际运行只绑定这些工具，不是再刷一遍字典索引。


````python
def build_tool_cabinet(source_reader: BaseTool) -> dict[str, BaseTool]:
    """为指定的真实资料工具创建用途明确的工具目录。

    Args:
        source_reader: 教师窄资料工具或已完成的本人工具。
    Returns:
        工具名到工具对象的字典，调用时才真正取正文。
    Raises:
        无：实际取件错误由工具调用返回或抛出。
    """

    def bind_one(name: str, description: str, source_id: str) -> BaseTool:
        @tool(name, description=description)
        async def fetch() -> dict[str, Any]:
            return await source_reader.ainvoke({"document_id": source_id})

        return fetch

    cabinet = {}
    for name, description, source_id in LOOKUPS:
        cabinet[name] = bind_one(name, description, source_id)
    return cabinet


teacher_cabinet = build_tool_cabinet(demo_document)
tool_catalog = [
    {"name": t.name, "description": t.description, "parameters": t.args}
    for t in teacher_cabinet.values()
]
print(json.dumps(tool_catalog, ensure_ascii=False, indent=2))
````

## 2. 先跑全目录基线

这次访客要核对“周六下午能否到馆报名”，需要开馆和报名两份资料。基线把四个工具都给模型；它不一定使用全部工具。观察实际调用，不能把“提供四个”写成“调用四次”。


````python
question = "我周六下午想来报名志愿者，几点能进馆、什么时候去哪儿报名？请查清再告诉我。"
full_agent = create_agent(
    model,
    tools=list(teacher_cabinet.values()),
    system_prompt=SYSTEM_PROMPT,
    middleware=[ModelCallLimitMiddleware(run_limit=5, exit_behavior="error")],
)
started = perf_counter()
async with asyncio.timeout(90):
    full_result = await full_agent.ainvoke({"messages": [HumanMessage(question)]})
full_seconds = perf_counter() - started
display(Markdown(full_result["messages"][-1].text))
save_record(
    "teacher-all-tools.json",
    {
        "question": question,
        "offered": list(teacher_cabinet),
        "elapsed_seconds": full_seconds,
        "trace": public_trace(full_result["messages"]),
    },
)
````

### 真实模型先选择，再执行

ToolChoice只给出能力名称，不给出原文。选择来自一次真实模型调用，必须再检查名称是否在实际registry、数量是否越界。模型写出一个好听的工具名，并不会让它自动存在。

下面用同一问题与同一批工具比较。选择器额外调用一次模型；后面的agent最多五次，两个阶段分别有等待上限。记录时不能漏算选择开销。


````python
class ToolChoice(BaseModel):
    """阿灯本次咨询需要携带的工具名称。"""

    names: list[str] = Field(
        max_length=2, description="只选择目录中本次需要的工具，最多两个；无需工具可为空"
    )


chooser = model.with_structured_output(ToolChoice, method="function_calling")
started = perf_counter()
async with asyncio.timeout(90):
    choice = await chooser.ainvoke(
        [
            SystemMessage(SYSTEM_PROMPT + "\n先只选择要携带的工具，不执行取件，不编造目录外名称。"),
            HumanMessage(question + "\n工具目录：" + json.dumps(tool_catalog, ensure_ascii=False)),
        ]
    )
    if len(set(choice.names)) != len(choice.names):
        raise ValueError("选择包含重复工具")
    for name in choice.names:
        if name not in teacher_cabinet:
            raise ValueError("选择了不存在的工具：" + name)
    selected_tools = [teacher_cabinet[name] for name in choice.names]
    small_agent = create_agent(
        model,
        tools=selected_tools,
        system_prompt=SYSTEM_PROMPT,
        middleware=[ModelCallLimitMiddleware(run_limit=5, exit_behavior="error")],
    )
    selected_result = await small_agent.ainvoke({"messages": [HumanMessage(question)]})
selected_seconds = perf_counter() - started
print("实际选择：", choice.names)
display(Markdown(selected_result["messages"][-1].text))
````

### 比较什么才诚实

schema_chars只是工具说明的字符数，不是token计费；elapsed_seconds是本次墙钟时间，包含网络波动。少传工具可能减少歧义，也可能因选择器漏选而变差。一次结果不能证明普遍优劣。

检查两次答复是否都真正取得NOTICE-B与VOL-01；只看语言相似不够。若短名单漏了资料，即使更快也不能叫优化。


````python
def used_tool_names(messages: list[AnyMessage]) -> list[str]:
    """从公开申请提取实际调用名。

    Args:
        messages: 实际运行消息。
    Returns:
        按发生顺序记录的工具名，可能重复。
    Raises:
        无：只读取标准工具字段。
    """
    used = []
    for message in messages:
        if isinstance(message, AIMessage):
            for call in message.tool_calls:
                used.append(call["name"])
    return used


comparison = {
    "question": question,
    "all": {
        "offered": list(teacher_cabinet),
        "used": used_tool_names(full_result["messages"]),
        "schema_chars": len(json.dumps(tool_catalog, ensure_ascii=False)),
        "seconds": full_seconds,
    },
    "selected": {
        "offered": choice.names,
        "used": used_tool_names(selected_result["messages"]),
        "selector_calls": 1,
        "seconds_including_selection": selected_seconds,
    },
    "trace": public_trace(selected_result["messages"]),
}
print(comparison["all"])
print(comparison["selected"])
save_record("teacher-selection-comparison.json", comparison)
````

```python
def used_tool_names(messages: list[AnyMessage]) -> list[str]:
    """从公开申请提取实际调用名。

    Args:
        messages: 实际运行消息。
    Returns:
        按发生顺序记录的工具名，可能重复。
    Raises:
        无：只读取标准工具字段。
    """
    used = []
    for message in messages:
        if isinstance(message, AIMessage):
            for call in message.tool_calls:
                used.append(call["name"])
    return used
```

输入 messages 是消息对象列表；遍历时只对 AIMessage 的 tool_calls 取 call["name"]，按发生顺序追加到 used 列表，允许重复。输出是工具名字符串列表。它不执行工具，只从已有消息中读取调用名。
