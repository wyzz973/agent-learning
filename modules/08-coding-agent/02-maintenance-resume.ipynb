### 分清三层职责

### 维修续做要同时保存目标、工作区与验证事实

长任务中，聊天记录、代码文件、允许改动范围和测试回执各自保存不同信息。只恢复对话不能知道工作区是否还是受检版本，只保留文件也不能确认哪一步曾经通过。持久维修记录需要把任务ID、目标、工作区身份、内容版本、已验证步骤和未完成事项连接起来；再次进入时先核对这些事实，再决定继续哪里。

Python生命周期管理负责启动、等待、取消和清理，幂等标识帮助避免重复副作用；检查点本身不替你持久保存容器文件。容器结束后工作区是否可恢复取决于明确的存储挂载和清理策略，不能把销毁资源与保留作品混说。程序化工具组合可能减少模型往返，但批量中间结果、失败位置和权限边界仍应可观察，不能为了更少调用而失去验证。迁移到两个组件的维护时，依赖步骤需要顺序，独立操作才适合组合；每项完成都对应当前版本的环境证据，而非沿用旧绿色记录。

````python
def public_trace(messages: list[AnyMessage]) -> list[dict[str, Any]]:
    """提取公开消息，完整AIMessage仍只在运行内存使用。

    Args:
        messages: 本次实际消息列表。
    Returns:
        角色、公开正文、工具调用及关联id，不含隐藏推理或请求元数据。
    Raises:
        无：只读取已知字段。
    """
    rows = []
    for message in messages:
        row: dict[str, Any] = {"role": message.type, "text": message.text}
        if isinstance(message, AIMessage):
            row["tool_calls"] = message.tool_calls
        if isinstance(message, ToolMessage):
            row["tool_call_id"] = message.tool_call_id
            row["tool_name"] = message.name
            row["status"] = message.status
        rows.append(row)
    return rows
````

````python
def save_record(name: str, payload: dict[str, Any]) -> Path:
    """保存本次实际记录，不推断学习者已掌握。

    Args:
        name: 当前目录中的JSON文件名，不允许路径片段。
        payload: 可序列化的公开数据，不包含密钥或隐藏推理。
    Returns:
        写入文件路径。
    Raises:
        ValueError: 文件名越界；OSError: 写入失败。
    """
    if Path(name).name != name or not name.endswith(".json"):
        raise ValueError("只允许当前作品目录中的JSON文件名")
    path = RUN_DIR / name
    path.write_text(json.dumps(payload, ensure_ascii=False, indent=2), encoding="utf-8")
    return path
````

## 1. 工单不是一份“曾经成功”的回忆

workspace标识允许操作的具体工作台；session标识一次持续委托；run是一次启动。一个session可以跨多个run，但不能换到别人的workspace。源码SHA把检查结果绑定到文件版本；相同任务标题不足以证明版本相同。

下面先让真实模型根据失败记录写一个短交接说明，不给它修改工具。交接仅是自然语言计划，不能替代实际修改或通过检查。


````python
demo_goal = "修复candidate.py的excerpt，让它准确返回Python半开区间[start,end)；固定测试不可改。"
async with asyncio.timeout(45):
    plan_reply = await model.ainvoke(
        [
            SystemMessage(SYSTEM_PROMPT),
            HumanMessage(
                "先替林禾留下简短维修交接，不修改文件。目标："
                + demo_goal
                + "\n当前源码："
                + DEMO_SOURCE
                + "\n检查结果："
                + demo_before["stderr"]
            ),
        ]
    )
display(Markdown(plan_reply.text))
demo_lab.set_write_permission(True)
demo_session = {
    "kind": "teacher-reference",
    "session_id": "maintenance-" + uuid.uuid4().hex[:12],
    "workspace": str(demo_lab.run_dir),
    "workspace_id": demo_lab.meta["workspace_id"],
    "goal": demo_goal,
    "allow_write": True,
    "allowed_files": ["candidate.py"],
    "candidate_sha": demo_lab.source_sha(),
    "tests_sha": demo_lab.meta["tests_sha"],
    "status": "paused",
    "completed_steps": ["read_source", "checked_initial_failure"],
    "handoff_note": plan_reply.text,
    "created_pid": os.getpid(),
    "latest_test": demo_before,
}
demo_session_file = RUN_DIR / "teacher-session.json"
demo_session_file.write_text(
    json.dumps(demo_session, ensure_ascii=False, indent=2), encoding="utf-8"
)
print("已暂停，没有把计划登记成修复：", demo_session_file)
````

### 本人恢复与程序化工具编排

resume_maintenance读取工单而不是Notebook残留变量。验证session、workspace边界、测试与候选SHA、允许文件和写入权限；版本漂移时抛ValueError，不能自动覆盖或继续使用旧检查。

重新创建三个工作台工具（read_code/edit_code/run_tests），再调用**本人run_guarded_repair**。这叫程序化编排：程序先验证记录和授权，再交给模型处理剩余动作，最后独立重测；不是让模型自由决定自己有没有权限。

返回session_id、workspace_id、status、candidate_sha、tests、messages。最近通过测试需匹配最终源码，agent最后一张测试回执也需对应当前SHA。若原候选已经通过，可以只重新核验；不要为了剧情故意宣称存在缺陷。

分层提示：先用新对象读工单；再对照谁能决定权限；最后才接模型。不能直接返回教师resumed或调用固定教师恢复脚本。


````python
async def resume_maintenance(record_file: Path, model: Any, system_prompt: str, allowed_root: Path) -> dict[str, Any]:
    """本人从磁盘工单恢复自己的维修控制，不依赖Notebook旧对象。

    Args:
        record_file: 本人工单；model: 真实模型；system_prompt: 本关岗位要求；allowed_root: 工作区边界。
    Returns:
        session_id/workspace_id/status/candidate_sha/tests/messages，实际验证结果。
    Raises:
        ValueError: 工单、权限、版本或路径不符；TimeoutError: 整段超过90秒；NotImplementedError: 未完成。
    """
    raise NotImplementedError("请本人实现验证工单、重建工具、调用本人控制与最终验证。")

print("下一段由全新进程首次真正执行本人恢复函数；未实现将明确失败。")
````

## 5. 拿走许可或改错版本，会怎样

只改工单副本，把candidate_sha换成明显不匹配的值，应在模型调用前拒绝。把allow_write改False则可以选择只读核对或明确blocked，但不可写入；用文件SHA和工具轨迹验证，而非只看答复说“我没改”。

迁移不是重写循环：为另一条本馆维护目标建立新session，指向独立副本，沿用本人恢复函数，不能借用原session的权限或绿色结果。


````python
bad_record = dict(my_record)
bad_record["candidate_sha"] = "0" * 64
bad_file = RUN_DIR / "stale-session.json"
bad_file.write_text(json.dumps(bad_record, ensure_ascii=False), encoding="utf-8")
try:
    await resume_maintenance(bad_file, model, SYSTEM_PROMPT, ROOT / "outputs/fog-island")
except ValueError:
    print("正确：版本不符时没有继续维修。")
else:
    raise AssertionError("工单版本过期，不能直接恢复。")
current_lab = CodingWorkspace.reopen(my_workspace, ROOT / "outputs/fog-island")
read_only_record = create_maintenance_record(current_lab, my_record["goal"], False)
read_only_file = RUN_DIR / "read-only-session.json"
read_only_file.write_text(json.dumps(read_only_record, ensure_ascii=False), encoding="utf-8")
sha_before_read_only = current_lab.source_sha()
try:
    read_only_result = await resume_maintenance(read_only_file, model, SYSTEM_PROMPT, ROOT / "outputs/fog-island")
except ValueError:
    print("当前恢复策略明确拒绝了未获写入授权的任务。")
else:
    assert read_only_result["status"] in {"blocked", "read_only", "verified", "needs_attention"}
assert current_lab.source_sha() == sha_before_read_only
````
